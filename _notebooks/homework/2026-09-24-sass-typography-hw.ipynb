## Homework Hack

The cat petting guide is rebuilt in an OCS card with semantic tags:

* Body text: `<p class="ocs__text">`
* The three steps: one `<ol>` with three `<li>`
* Title: `<h2 class="ocs__section-title">`
* Important text: `<strong>`
* Section headings: `<h3>`

{% capture challenge2 %}
Homework Hack: the cat guide with OCS classes, an ordered list, strong text, and semantic tags. Hit Run.
{% endcapture %}

{% capture code2 %}
// Update the HTML inside the quotes, then hit Run.
outputElement.innerHTML = [
  '<div class="ocs__card">',
  '  <h2 class="ocs__section-title">A Beginner Guide to Petting a Cat</h2>',
  '  <h3>Curious about how to…</h3>',
  '  <p class="ocs__text">pet a cat? Great news, we will walk you through it one step at a time!</p>',
  '  <p class="ocs__text"><strong>Cat petting steps:</strong></p>',
  '  <ol>',
  '    <li>let the cat sniff your hand</li>',
  '    <li>rest your hand on its back</li>',
  '    <li>stroke gently from head to tail</li>',
  '  </ol>',
  '  <h3>Well done! The cat is happy :D</h3>',
  '</div>'
].join('');

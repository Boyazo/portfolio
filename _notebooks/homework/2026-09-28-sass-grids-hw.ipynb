## Popcorn Hack: Refactor the Flex Row into an OCS Grid

The inline `display: flex` and the made-up `box` class are replaced with one grid root (`ocs__grid`), one variant (`ocs__grid--standard`), and three grid cells (`ocs__grid-cell`).

```python
%%html

<!-- UI_RUNNER: Grids Popcorn Base -->

<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">HTML</div>
  <div class="ocs__grid-cell">CSS</div>
  <div class="ocs__grid-cell">JS</div>
</div>
```

**Why this works:** `ocs__grid` is always the outer element, exactly one variant class picks the layout, and every direct child is an `ocs__grid-cell`. The standard variant is 3 columns (2 on mobile) and follows the site theme, so no CSS is needed.

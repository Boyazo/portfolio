## Popcorn Hack: Semantic Refactor

The `<style>` block and the three custom classes are gone. Meaning comes from the tags (`h1`, `h2`, `p`) and the look comes from `ocs_` classes: `ocs_section-title` for the main heading, `ocs_lead` for the supporting heading, and `ocs_text` for the paragraph.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Base-->

<h1 class="ocs_section-title">This is</h1>
<h2 class="ocs_lead">a very cool</h2>
<p class="ocs_text">lesson</p>
```

**Why:** A `<div class="large-title">` looks like a title but tells a screen reader nothing. `<h1>` says "this is the page title", and the theme already styles it, so the custom CSS is not needed.

## Homework Hack: How to Pat a Cat

- Fake headings (`title-font`, `section-font`) become real headings: `<h2>` for the title and `<h3>` for the section headings, so the hierarchy goes down one level at a time.
- `body-text` divs become `<p>` paragraphs.
- The visually bold `very-important` span becomes `<strong>`.
- The three fake numbered lines become **one** real `<ol>` with three `<li>` items (the browser numbers them, so the "1." "2." "3." typed by hand are removed).
- Everything is grouped in an `ocs_card`, and the title uses `ocs_section-title`.

```python
%%html
<!-- UI_RUNNER: Typography Homework Base -->

<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to…</h3>
  <p class="ocs_text">pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>put out your hand</li>
    <li>touch the cat</li>
    <li>move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

**Checklist:**

- `%%html` and the `UI_RUNNER` comment are kept.
- No `<style>` block, no inline styles, no custom classes, no `<span>` fake emphasis.
- Heading tags (`h2`, `h3`), paragraph tags, one ordered list with three list items, and `<strong>` for the previously "very important" text.
- `ocs_` classes used: `ocs_card`, `ocs_section-title`, `ocs_text`.

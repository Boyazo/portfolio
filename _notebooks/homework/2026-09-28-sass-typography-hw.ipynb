### Homework Hack: How to Pat a Cat

- Fake headings become real headings: `<h2>` for the title and `<h3>` for the section headings, going down one level at a time.
- `body-text` divs become `<p>` paragraphs.
- The visually bold `very-important` span becomes `<strong>`.
- The three hand-numbered lines become **one** real `<ol>` with three `<li>` items, so the browser does the numbering.
- Everything is grouped in an `ocs_card`, and the title uses `ocs_section-title`. No `<style>` block, inline styles, or custom classes.

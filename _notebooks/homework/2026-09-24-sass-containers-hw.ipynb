## Popcorn Hack: Parade Invite Makeover

One `ocs__card` inside one `ocs__container` holds the full Poway Scripps Rotary Parade invite. The card starts with an `ocs__section-title` and a one-sentence `ocs__description`, then a `cols-3` standard grid lists the schedule one item per cell. The start time is saved for last and marked as the `--accent` cell.

## Homework Hack: Parade Home Base

The Poway Rotary Parade home base page has:

* A final `ocs__callout` line
* An `ocs__section-title` and a one-sentence `ocs__description`
* One `ocs__container` wrapped around one `ocs__card`
* A `cols-3` standard grid containing a `--muted` cell, an `--accent` cell, and a `--header` cell
* An `ocs__table-wrap` + `ocs__table` table with a header row and two body rows

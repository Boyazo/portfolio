## Popcorn Hack: Class-Based Inputs

The inline width/border and the made-up `box` class are replaced with `ocs__input`: one **large gradient** search field and one **small** plain field.

```python
%%html
<!-- UI_RUNNER: Inputs Popcorn Base -->

<input type="text" class="ocs__input large gradient" placeholder="Search PVO resources..." aria-label="Search PVO resources">
<input type="text" class="ocs__input small" placeholder="Volunteer First Name" aria-label="Volunteer first name">
```

**Why:** `ocs__input` is the base role, `large`/`small` set the size, and `gradient` replaces the hand-written dashed purple border. Modifiers can be combined (`large gradient`).

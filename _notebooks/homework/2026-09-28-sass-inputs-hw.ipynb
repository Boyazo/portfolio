### Homework Hack: Refactor the Signup Form

- The 500px pink-to-purple field becomes `ocs__input large gradient`.
- `tinybox` becomes `ocs__input small`.
- The email field's inline padding becomes `ocs__input medium`, and it keeps `type="email"` so browsers validate it.
- Each field keeps a clear purpose through its placeholder, and an `aria-label` gives it an accessible name.

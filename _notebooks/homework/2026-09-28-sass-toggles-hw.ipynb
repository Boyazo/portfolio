## Homework Hack: Study Settings Panel

Four toggles, each with its own purpose and label. JavaScript reads `.checked` on each one and updates the page, a live counter shows how many are on, and three of the toggles are connected to a real feature on the page (a shown/hidden tip, a dark preview box, and larger text). Comments throughout explain what I learned.

```python
%%html
<!-- UI_RUNNER: Toggles Homework -->

<div id="toggle-demo">
  <label class="ocs__toggle ocs__toggle--green">
    <input id="tip-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show study tip</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="dark-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Dark preview</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="big-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Large text</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="remind-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Homework reminders</span>
  </label>

  <p id="tip" hidden>Tip: study for 25 minutes, then take a 5 minute break.</p>
  <div id="preview">This preview box reacts to the toggles above.</div>
  <p id="status"></p>
  <p id="count" aria-live="polite"></p>
</div>

<script>
  // I learned that .checked lets JavaScript tell whether a toggle is currently on or off.
  // The checkbox keeps the state, so I never have to track true/false myself.
  const tipToggle = document.getElementById("tip-toggle");
  const darkToggle = document.getElementById("dark-toggle");
  const bigToggle = document.getElementById("big-toggle");
  const remindToggle = document.getElementById("remind-toggle");

  const tip = document.getElementById("tip");
  const preview = document.getElementById("preview");
  const status = document.getElementById("status");
  const count = document.getElementById("count");

  // One function redraws everything so the page always matches the toggles
  function update() {
    // Toggle 1 controls a feature: show or hide the tip (hidden = NOT checked)
    tip.hidden = !tipToggle.checked;

    // Toggle 2 controls a feature: dark or light preview colors
    preview.style.background = darkToggle.checked ? "#1f2937" : "#f3f4f6";
    preview.style.color = darkToggle.checked ? "#ffffff" : "#111827";
    preview.style.padding = "12px";

    // Toggle 3 controls a feature: larger or normal text
    preview.style.fontSize = bigToggle.checked ? "24px" : "16px";

    // Toggle 4 changes a status message
    status.textContent = remindToggle.checked ? "Reminders are ON" : "Reminders are OFF";

    // Count how many toggles are on. This is a selection question for each one:
    // IF the toggle is checked THEN add 1, which is exactly a conditional from AP CSP.
    const toggles = document.querySelectorAll("#toggle-demo .ocs__toggle-input");
    let on = 0;
    toggles.forEach(function (t) {
      if (t.checked) {
        on += 1;
      }
    });
    count.textContent = "Toggles turned on: " + on + " of " + toggles.length;
  }

  // The "change" event fires every time a toggle flips, so the counter updates live
  [tipToggle, darkToggle, bigToggle, remindToggle].forEach(function (t) {
    t.addEventListener("change", update);
  });

  // Run once at load so the page starts with the correct text (0 of 4 on)
  update();
</script>
```

**How it meets each requirement:**

- **At least 3 toggles, each with its own label and purpose:** 4 toggles (study tip, dark preview, large text, homework reminders).
- **JavaScript checks each on/off state:** `.checked` is read on every toggle inside `update()`.
- **Section showing how many are on, updating live:** `Toggles turned on: N of 4`, refreshed by the `change` event.
- **At least 2 comments showing understanding:** comments explain `.checked`, why one `update()` function redraws everything, and that counting checked toggles is an AP CSP conditional.
- **Above and beyond:** three toggles trigger real features on the page (show/hide the tip, dark preview, large text) instead of only changing their own appearance.

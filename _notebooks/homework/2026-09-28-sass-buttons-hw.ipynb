## Homework Hack #1: Adaptive Resource Toolbar

All three actions are wrapped in `ocs__links`, every inline `style` and hex color is removed, and each category gets the classes the requirements specify. The starter's `href` values were also broken (Markdown link syntax pasted inside the attribute), so they are fixed to plain URLs.

```python
%%html
<!-- UI_RUNNER: Buttons Homework 1 -->

<div class="ocs__links">
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-red fill pill">Urgent Support</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-green fill pill">Self-Care</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn accent outline pill">Community</a>
</div>
```

**Why these classes:** red + `fill` is the loudest option for urgent support, green + `fill` signals a positive self-care action, and accent + `outline` is the quieter option for community. All three use `<a>` because they navigate to a resource page.

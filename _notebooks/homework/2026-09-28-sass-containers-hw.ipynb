- The table rebuilt with `ocs__table-wrap` + `ocs__table` (one header row and two body rows).

### Optional SASS Extra (+0.2): A Tinted Container Variant

Kept **separate** from the graded page so the required part stays at zero custom CSS:

```scss
// Tinted variant of the shared page container.
.ocs__container.my-container--tinted {
  background: rgba(34, 197, 94, 0.08); // soft green tint behind the page content
  border-radius: 12px;                 // rounded edges like the cards inside it
  max-width: 1000px;                   // a little wider than the default 900px
}
```

```html
<div class="ocs__container my-container--tinted">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Hub</h2>
  </div>
</div>
```

The class sits *next to* `ocs__container`, so it keeps every shared rule (centering, padding, the user's font and text color) and only overrides what it needs to.
